## Capa Stream - Datos fluidos (hipotético)

```python
## ciudades con baja representación:
conteo_paises = df["country"].value_counts()

try:
    
    for _, fila in df.iterrows():

        time.sleep(1)
    
        pais = fila["country"]
        ciudad = fila["name"]

        if conteo_paises[pais] < 3:

            print(
                f"""
    LOW REPRESENTATION ALERT

    City: {ciudad}
    Country: {pais}

    Only {conteo_paises[pais]} city(s) from this country
    are present in the historical dataset.
"""
        )
```
