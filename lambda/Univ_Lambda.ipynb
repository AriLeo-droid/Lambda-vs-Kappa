## Capa streaming - Datos continuos y alertas (hipotético)

```python
try:
    for _, fila in df.iterrows():

        time.sleep(1)
        
        if fila["Hlab"] > 4:
            
            print(f"Alerta: {fila['Asignatura']} tiene carga alta")
            
except KeyboardInterrupt:
    print("Streaming detenido")
```

```python
prom_creditos = df["Cred"].mean()
try:
    for _, fila in df.iterrows():
        
        time.sleep(1)
        
        if fila["Cred"] > prom_creditos:
            
            print(f"ALERTA: {fila['Asignatura']} supera la carga promedio.")
            
except KeyboardInterrupt:
    print("Streaming detenido")
```
