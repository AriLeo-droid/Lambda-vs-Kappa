## Capa Streaming - Datos fluidos, alertas (hipotético)

```python
try:
    for _, fila in df.iterrows():
        
        time.sleep(1)
        
        rating = fila["Hospital overall rating"]

        if pd.notna(rating) and float(rating) <= 2:

            print(f"""
    QUALITY ALERT!!

    Provider ID: {_}
    Hospital: {fila['Hospital Name']}
    State: {fila['State']}
    Rating: {rating}
    """
            )

except KeyboardInterrupt:
    print("Streaming detenido")
```

```python
try:
    for _, fila in df.iterrows():
        
        time.sleep(1)
        
        if fila["Safety of care national comparison"] == \
           "Below the National average":

            print(
                f"""
    SECURITY ALERT!!

    Provider ID: {_}
    Hospital: {fila['Hospital Name']}
    City: {fila['City']}
    State: {fila['State']}
    """
            )
except KeyboardInterrupt:
    print("Streaming detenido")
```

```python
try:
    for _, fila in df.iterrows():
        
        time.sleep(1)
        
        if fila["Emergency Services"] == False:

            print(
                f"""
    OPERATIVE ALERT!!

    Provider ID: {_}
    Hospital: {fila['Hospital Name']}

    Does not provide emergency services.
    """
            )
except KeyboardInterrupt:
    print("Streamnig detenido")
```
