```python
import pandas as pd
import numpy as np
import seaborn as sns
import missingno as msno
import matplotlib.pyplot as plt
import time
```

```python
df = pd.read_csv("..\\data\\Ciudades\\Cities.csv", index_col="id")
df.reset_index(drop=True, inplace=True)
df
```

```python
df.shape
```

```python
# Aleatorización de datos
df = (
    df.sample(
        frac=1,
        random_state=42
    )
    .reset_index(drop=True)
)
```

## Arquitectura Lambda - datos fluidos

```python
ciudades_por_pais = {}
try:
    for _, fila in df.iterrows():

        pais = fila["country"]

        ciudades_por_pais[pais] = (
            ciudades_por_pais.get(pais, 0) + 1
        )
        time.sleep(1)
        if ciudades_por_pais[pais] < 3:

            print(
                f"""
    LOW REPRESENTATION ALERT

    City: {fila['name']}
    Country: {pais}

    Current Records:
    {ciudades_por_pais[pais]}
    """
            )
except KeyboardInterrupt:
    print("streaming detenido")
```
