## Arquitectura Kappa - datos dinámicos sin historial

```python
## Promedio de créditos de materias conforme más son implementadas
suma_creditos = 0
contador = 0

try:
    for _, fila in df.iterrows():

        contador += 1
        suma_creditos += fila["Cred"]

        promedio_actual = suma_creditos / contador

        time.sleep(1)

        print(
            f"Promedio actual: {promedio_actual:.2f}"
        )
except KeyboardInterrupt:
    print("Streaming detenido")
```

```python
## Materia más exigente (dinámico)
max_horas = 0
try:
    for _, fila in df.iterrows():

        horas = fila["Hclas"] + fila["Hlab"]

        if horas > max_horas:

            max_horas = horas

            time.sleep(1)
            
            print(
                f"""
    NUEVO RÉCORD

    Materia: {fila['Asignatura']}
    Horas totales: {horas}
    """
        )
except KeyboardInterrupt:
    print("Streaming detenido")
```
