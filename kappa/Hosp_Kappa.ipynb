## Arquitectura Kappa - Streaming constante de archivos

```python
suma_rating = 0
contador = 0
try:
    for provider_id, fila in df.iterrows():

        rating = fila["Hospital overall rating"]

        if pd.notna(rating):
    
            rating = float(rating)

            contador += 1
            suma_rating += rating

            promedio_actual = suma_rating / contador

            time.sleep(1)
            
            if rating < promedio_actual:

                
                print(
                    f"""
    QUALITY ALERT

    Provider ID: {provider_id}
    Hospital: {fila['Hospital Name']}

    Current Rating: {rating}
    Current Stream Average: {promedio_actual:.2f}
    """
                )
except KeyboardInterrupt:
    print("Streaming detenido")
```
